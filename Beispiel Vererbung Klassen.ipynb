Importieren wichtiger Libaries 

```python
from datetime import date
```

Verbende Klasse (superclass) wird definiert

```python
class Transportmittel:
    def __init__(self, name: str, gewicht: float, farbe: list[int, int, int], kaufdatum: date):
        self.name: str = name
        self.gewicht: float = gewicht
        self.farbe: list = farbe
        self.kaufdatum: date = kaufdatum

    def print_superclass_details(self):
        print(f"Name:{self.name}")
        print(f"Gewicht:{self.gewicht} kg")
        print(f"Farbe RGB-Code:{self.farbe}")
        print(f"Kaufdatum: {self.kaufdatum}")
        print("")
```

Erbende Klassen (subclass) wird definiert

```python
class Bike(Transportmittel):
    def __init__(self, name: str, gewicht: float, farbe: list[int, int, int], reifengroeße: int, kaufdatum: date, fahrradtype: str ):
        super().__init__(name, gewicht, farbe,  kaufdatum)
        
        self.reifengroeße: int = reifengroeße
        self.fahrradtype: str = fahrradtype

    def print_subclass_details(self):
        print(f"Reifengroeße: {self.reifengroeße} zoll")
        print(f"Fahrradtype: {self.fahrradtype}")
        print("")

class Auto(Transportmittel):
    def __init__(self, name: str, gewicht: float, farbe: list[int, int, int], ps: int, kaufdatum: date, fahrzeug_type: str ):
        super().__init__(name, gewicht, farbe, kaufdatum)
        
        self.ps: int = ps
        self.fahrzeug_type: str = fahrzeug_type

    def print_subclass_details(self):
        print(f"PS:{self.ps} zoll")
        print(f"fahrzeug_type: {self.fahrzeug_type}")
        print("")
```



Initialisieren von 2 Objekten (1x Bike; 1x Auto)

```python
bike_object_one = Bike("Laras Fahrrad", 2.1, [255, 0, 0], 24, date(2021, 3, 12), "Rennrad")
auto_object_two = Auto("Lucas Auto", 872.3, [255, 255, 0], 15, date(2017, 3, 12), "VW UP")
```

die Methode print_details der beiden Objekte aufrufen

```python
bike_object_one.print_superclass_details()
bike_object_one.print_subclass_details()
```

```python
auto_object_two.print_superclass_details()
auto_object_two.print_subclass_details()
```
