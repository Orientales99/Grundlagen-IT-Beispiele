Klasse Rechner wird erstellt mit einer Methode addieren()

```python
class Rechner:
    def addiere(self, a=None, b=None, c=None):
        if a is not None and b is not None and c is not None:
            return a + b + c
        elif a is not None and b is not None:
            return a + b
        elif a is not None:
            return a
        else:
            return 0
```

Die Methode wird mit jeweils einer unterschiedlichen Anzahl an Parametern aufgerufen, wobei dadaurch andere Teile des Codes verwendet werden 

```python
r = Rechner()
print(r.addiere(2))        # ➜ 2
print(r.addiere(2, 3))     # ➜ 5
print(r.addiere(2, 3, 4))  # ➜ 9
```
