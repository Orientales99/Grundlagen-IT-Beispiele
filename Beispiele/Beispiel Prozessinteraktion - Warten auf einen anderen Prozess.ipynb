```python
from random import seed, randint

seed(23)


import simpy
```

Die Klasse ElektronischesFahrzeug wird definiert. 
Das Fahrzeug fährt erst wieder los, wenn die Zeit vom parking und von der batterie_steuerung() durchgelaufen sind. Dabei laufen parking und charging parallel ab. 

```python
class ElektronischesFahrzeug:
    def __init__(self, env: simpy.Environment):
        self.env = env
        self.drive_proc = env.process(self.drive(env))

    def drive(self, env: simpy.Environment):
        """Das Fahrzeug fährt 20-40 min und startet danach das Parken für 1-6 h. 
        Während des parkens wird die Methode batterie_steuerung aufgerufen. Es wird mit Hilfe
        des yield-Befehls gewartet, bis charging und parking durchgelaufen sind."""
        while True:
            # Fahre für 20-40 min
            yield env.timeout(randint(20,40))

            # Parke für 1-6 h
            print("Starte das parken um: ", env.now)
            charging = env.process(self.batterie_steuerung(env))
            parking = env.timeout(randint(60, 360))
            yield charging & parking
            print("Parken endet um:", env.now)

    def batterie_steuerung(self, env: simpy.Environment):
            print("Batterie laden startet um:", env.now)
            yield env.timeout(randint(30, 90))
            print("Batterie wurde fertig geladen um:", env.now)


env = simpy.Environment()
ev = ElektronischesFahrzeug(env)

env.run(until=150)
```
