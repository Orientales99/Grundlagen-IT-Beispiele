Die benötigten Bibliotheken werden importiert

```python
import simpy
```

Der Generator produzent() wird definiert

```python
def produzent(env: simpy.Environment, store: simpy.Store):
    for i in range(100):
        yield env.timeout(2)
        yield store.put(f'Produkt {i}')
        print(f'Hat Produkt produziert um ', env.now)
```

Der Generator konsument() wird definiert

```python
def konsument(name: str, env: simpy.Environment, store: simpy.Store):
    while True:
        yield env.timeout(1)
        print(name, 'hat Produkt angefordert um ', env.now)
        item = yield store.get()
        print(name, "bekommt ", item, "um ", env.now)
```

Das Environment() wir erstellt und ein store wird definiert. Außerdem werden die Prozesse produzent & konsument gestartet

```python
env = simpy.Environment()
store = simpy.Store(env, capacity=2)

prod = env.process(produzent(env, store))
consumers=[env.process(konsument(f'Kunde {i}', env, store)) for i in range(2)]
```
