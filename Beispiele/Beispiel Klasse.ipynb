Importieren wichtiger Libaries 

```python
from datetime import date
```

Klasse Bike wird definiert

```python
class Bike:
    def __init__(self, name: str, gewicht: float, farbe: list[int, int, int], reifengröße: int, kaufdatum: date, fahrradtype: str ):
        self.fahrrad_name: str = name
        self.gewicht: float = gewicht
        self.farbe: list = farbe
        self.reifengroeße: int = reifengröße
        self.kaufdatum: date = kaufdatum
        self.fahrradtype: str = fahrradtype

    def print_details(self):
        print(f"Name:{self.fahrrad_name}")
        print(f"Gewicht:{self.gewicht} kg")
        print(f"Farbe RGB-Code:{self.farbe}")
        print(f"Reifengroeße:{self.reifengroeße} zoll")
        print(f"Kaufdatum: {self.kaufdatum}")
        print(f"Fahrradtype: {self.fahrradtype}")
        print("")
```

Initialisieren von 2 Objekten der Klasse Bike

```python
bike_object_one = Bike("Laras Fahrrad", 2.1, [255, 0, 0], 24, date(2021, 3, 12), "Rennrad")
bike_object_two = Bike("Bens Fahrrad", 3.3, [255, 255, 0], 20, date(2021, 3, 12), "Stadtrad")
```

die Methoden print_details der beiden Objekte werden aufgerufen

```python
bike_object_one.print_details()
bike_object_two.print_details()
```
