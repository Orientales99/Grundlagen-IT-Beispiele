Verbende Klasse (superclass) wird definiert

```python
from datetime import date
```

Verbende Klasse (superclass) wird definiert

```python
class Produktionsroboter:
    def __init__(self, id, energieverbrauch, kaufdatum):
        self.id: str = id
        self.energieverbrauch: float = energieverbrauch
        self.kaufdatum: date = kaufdatum
        self.beschaeftigt_status: bool = False
```

Erbende Klassen (subclass) wird definiert

```python
class Maschine(Produktionsroboter):
    def __init__(self, id, energieverbrauch, kaufdatum, produktionsgeschwindigkeit, benoetigt_arbeitsroboter):
        super().__init__(id, energieverbrauch, kaufdatum)
        self.produktionsgeschwindigkeit: int = produktionsgeschwindigkeit
        self.benoetigt_arbeitsroboter: bool = benoetigt_arbeitsroboter

    def print_details(self):
        print(f"Id: {self.id}")
        print(f"Energieverbrauch {self.energieverbrauch} kw/h")
        print(f"Beschäftigt: {self.beschaeftigt_status}")
        print(f"Produktionsgeschwindigkeit: {self.produktionsgeschwindigkeit} Einheiten/Minute")
        print(f"Wird ein Arbeitsroboter benötigt: {self.benoetigt_arbeitsroboter}")
        print("")


class Arbeitsroboter(Produktionsroboter):
    def __init__(self, id, energieverbrauch, kaufdatum, liste_erfuellte_arbeitsauftraege = []):
        super().__init__(id, energieverbrauch, kaufdatum)
        self.liste_erfuellte_arbeitsauftraege: list[str, int] = liste_erfuellte_arbeitsauftraege

    def print_details(self):
        print(f"Id: {self.id}")
        print(f"Energieverbrauch {self.energieverbrauch} kw/h")
        print(f"Beschäftigt: {self.beschaeftigt_status}")
        print(f"Aufträge wurden fertiggestellt: {self.liste_erfuellte_arbeitsauftraege}")
        print("")
```


Initialisieren von 5 Objekten (2x Arbeitsroboter; 3x Maschine)

```python
arbeitsroboter_one = Arbeitsroboter("AR1", 0.5, date(2021, 3, 12))
arbeitsroboter_two = Arbeitsroboter("AR2", 0.5, date(2021, 3, 12))

teigmaschine = Maschine("TGM1", 60, date(2021, 3, 12), 1, True)
schokoladenaufbereitungsmaschine = Maschine("SAM1", 30, date(2019, 2, 1), 1, False)
ueberzugmaschine = Maschine("UZM1", 9, date(2021, 3, 12), 1, True)
```

die Methode print_details aller Objekte aufrufen

```python
arbeitsroboter_one.print_details()
arbeitsroboter_two.print_details()

teigmaschine.print_details()
schokoladenaufbereitungsmaschine.print_details()
ueberzugmaschine.print_details()
```
